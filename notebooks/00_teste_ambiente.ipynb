```python
import bs4
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import requests
import seaborn as sns

print("=" * 45)
print("  VALIDAÇÃO DO AMBIENTE: PROJETO UFC DATA")
print("=" * 45)

# 1. Teste de versões das bibliotecas instaladas
modulos = {
    "Requests (Coleta Web)": requests.__version__,
    "BeautifulSoup (HTML Parser)": bs4.__version__,
    "Pandas (Manipulação)": pd.__version__,
    "NumPy (Matrizes e Vetores)": np.__version__,
    "Matplotlib (Motor de Gráficos)": plt.matplotlib.__version__,
    "Seaborn (Design Estatístico)": sns.__version__,
}

for lib, versao in modulos.items():
    print(f"✔ {lib.ljust(30)}: v{versao}")

# 2. Teste funcional prático: Requisição e Criação de DataFrame
url_teste = "http://ufcstats.com"
resposta = requests.get(url_teste, timeout=5)

if resposta.status_code == 200:
    print(
        f"\n✔ Conexão HTTP com UFCStats bem-sucedida! (Status: {resposta.status_code})"
    )
else:
    print(f"\n⚠ Atenção: Resposta inesperada do site: {resposta.status_code}")

# 3. Teste rápido de renderização de gráfico com estilo escuro (estilo da página)
dados_teste = pd.DataFrame(
    {
        "Lutador": ["Lutador Vermelho", "Lutador Azul"],
        "Golpes": [82, 64],
    }
)

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(6, 3), dpi=100)
cores = ["#E53935", "#1E88E5"]

sns.barplot(data=dados_teste, x="Lutador", y="Golpes", palette=cores, ax=ax)
ax.set_title("Teste Visual - Volume de Golpes", fontsize=11, weight="bold")
ax.set_ylabel("Total Conectado")
ax.set_xlabel("")
plt.tight_layout()
plt.show()

print("\n🚀 Ambiente 100% configurado e pronto para a raspagem!")
```
